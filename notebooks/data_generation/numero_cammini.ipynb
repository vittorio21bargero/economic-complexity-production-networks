```python
import sys
import numpy as np
import pandas as pd
import scipy.sparse as sp
from pathlib import Path

ROOT = next(p for p in [Path().resolve(), *Path().resolve().parents] if (p / "src").is_dir())
sys.path.insert(0, str(ROOT))

from src.C_alberizzazione import make_C_nilpotente
```

- RISOLUZIONE: 4 digits
- NILPOTENZA : NO

```python
C_4 = pd.read_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\data\HS02_all_years\4_digits\alligned\binarized\C\C_4_digits_HS02.parquet')
C_4_sparse = sp.csr_matrix(C_4.values.astype(float))
```

```python
result = {}
```

```python
# inizializzo v:= vettore dei cammini arrivanti in p
v = np.ones(C_4_sparse.shape[0])
for k in range(1, 10):
    v = v @ C_4_sparse
    result[f"numero_input_ordine_{k}"] = np.asarray(v).ravel()


df_numero_input = pd.DataFrame(result, index=C_4.columns)
```

```python
# numero di cammini entranti per prodotto (usato da plotting/numero_cammini.ipynb)
df_numero_input.rename_axis("prodotto").reset_index().to_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\results\numero_cammini\cammini_per_product\C_classica\cammini_per_product_4_digits.parquet')
```

```python
# matrice di correlazione 9x9
correlation_matrix = df_numero_input.corr(method="spearman")   # oppure method="spearman"
correlation_matrix.to_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\results\numero_cammini\correlation_matrix\C_classica\spearman_4_digits.parquet')   
```
