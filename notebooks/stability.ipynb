```python
import numpy as np
import pandas as pd
import sys
from pathlib import Path
from scipy.stats import pearsonr, spearmanr

ROOT = Path().resolve().parent
sys.path.insert(0, str(ROOT))

from src.algorithms import compute_extended_fitness
from src.mixed_functions import autovettore_principale, indegree_medio, link_rewiring, compute_alpha

```

```python
Mcp_4_digits = pd.read_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\data\HS02_all_years\4_digits\Mcp_export_alligned\Mcp_export_alligned_Y2002.parquet')
C_4_digits = pd.read_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\data\HS02_all_years\4_digits\C\C_4_digits_HS02.parquet')
```

```python
alpha = compute_alpha(C_4_digits)
```

```python
complexity, fitness , _ , _ = compute_extended_fitness(
            Mcp_4_digits, C_4_digits.T, alpha = alpha, max_iter = 10000,
            tol = 1e-6, track = False ,save_delta = False, delta_window = None)
```

```python
etas = np.linspace(0,1,11)
```

```python
rows = []   # creo una lista per costruire la tabella

for eta in etas:
    C_rewired = link_rewiring(C_4_digits, eta=eta, seed=1)

    alpha_rewired = compute_alpha(C_rewired)

    complexity_rewired, fitness_rewired, _, _ = compute_extended_fitness(
        Mcp_4_digits, C_rewired.T, alpha=alpha_rewired, max_iter=10000,
        tol=1e-6, track=False, save_delta=False, delta_window=None)

    pearson_coeff_Q, _ = pearsonr(complexity, complexity_rewired)
    spearman_coeff_Q, _ = spearmanr(complexity, complexity_rewired)
    pearson_coeff_F, _ = pearsonr(fitness, fitness_rewired)
    spearman_coeff_F, _ = spearmanr(fitness, fitness_rewired)

    # raggio spettrale della matrice rewired
    raggio_spettrale, _  = autovettore_principale(C_rewired.T)

    rows.append({
        "eta":              eta,
        "pearson_Q":        pearson_coeff_Q,
        "spearman_Q":       spearman_coeff_Q,
        "pearson_F":        pearson_coeff_F,
        "spearman_F":       spearman_coeff_F,
        "raggio_spettrale": raggio_spettrale,
    })

C_stability_table_4_digits = pd.DataFrame(rows)
```

```python
C_stability_table_4_digits.to_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\results\stability\C_stability_4_digits.parquet')
```

```python
import numpy as np
import pandas as pd
import sys
from pathlib import Path
from scipy.stats import pearsonr,spearmanr

ROOT = Path().resolve().parent
sys.path.insert(0, str(ROOT))

from src.algorithms import compute_extended_fitness
from src.mixed_functions import autovettore_principale, flip_Mcp, compute_alpha
```

```python
Mcp = pd.read_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\data\HS02_all_years\4_digits\Mcp_export_alligned\Mcp_export_alligned_Y2002.parquet')
C = pd.read_parquet(r'C:\Users\vitto\Desktop\CSH RESEARCH\data\HS02_all_years\4_digits\C\C_4_digits_HS02.parquet')
```

```python
alpha = compute_alpha(C)
```

```python
complexity, fitness , input_complexity , intrinsic_complexity = compute_extended_fitness(
            Mcp, C.T, alpha = alpha, max_iter = 10000,
            tol = 1e-6, track = False ,save_delta = False, delta_window = None)
```

```python
etas = np.linspace(0,1,101)
```

```python
rows = []

for eta in etas:
    Mcp_flipped = flip_Mcp(Mcp, eta = eta, seed = 50 )
    rho, _ = autovettore_principale(C.T)
    alpha = 1/rho
    complexity_flipped, fitness_flipped , input_complexity_flipped , intrinsic_complexity_flipped = compute_extended_fitness(
            Mcp_flipped, C.T, alpha = alpha, max_iter = 10000,
            tol = 1e-6, track = False ,save_delta = False, delta_window = None)
    
    pearson_coeff_Q, _ = pearsonr(complexity, complexity_flipped)
    spearman_coeff_Q, _ = spearmanr(complexity, complexity_flipped)
    pearson_coeff_F, _ = pearsonr(fitness, fitness_flipped)
    spearman_coeff_F, _ = spearmanr(fitness, fitness_flipped)
    rows.append({
        "eta":              eta,
        "pearson_Q":        pearson_coeff_Q,
        "spearman_Q":       spearman_coeff_Q,
        "pearson_F":        pearson_coeff_F,
        "spearman_F":       spearman_coeff_F,
        
    })

Mcp_stability_table_4_digits = pd.DataFrame(rows)    
```
