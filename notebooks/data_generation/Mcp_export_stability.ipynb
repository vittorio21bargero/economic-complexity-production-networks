```python
import numpy as np
import pandas as pd
import sys
from pathlib import Path
from scipy.stats import pearsonr, spearmanr

ROOT = next(p for p in [Path().resolve(), *Path().resolve().parents] if (p / "src").is_dir())
sys.path.insert(0, str(ROOT))

from src.algorithms_with_imports import compute_extended_fitness_with_imports, compute_extended_eci_with_imports
from src.C_properties import autovettore_and_autovalore_principale 
from src.algorithms import compute_alpha
from src.matrix_perturbations import Mcp_perturbation
```

- ALGORITMO: extended_fitness_with_imports
- RISOLUZIONE: 4 digits

```python
Mcp_export_4 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/4_digits/alligned/binarized/Mcp_export/Mcp_export_alligned_Y2002.parquet'))
Mcp_import_4 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/4_digits/alligned/binarized/Mcp_import/Mcp_import_alligned_Y2002.parquet'))
C_4 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/4_digits/alligned/binarized/C/C_4_digits_HS02.parquet'))
```

```python
alpha = compute_alpha(C_4)

complexity, fitness , input_complexity , intrinsic_complexity, import_penalty = compute_extended_fitness_with_imports(
            Mcp_export_4, Mcp_import_4, C_4.T, alpha = alpha, tensor_window = 3, max_iter = 1000,
            tol = 1e-6, track_convergence = False ,save_delta = False, delta_window = None)
```

```python
etas = np.linspace(0,1,11)
rows = []   # creo una lista per costruire la tabella


for eta in etas:
    
    Mcp_export_4_perturbed = Mcp_perturbation(Mcp_export_4, eta = eta, seed = 1)

    complexity_perturbed, fitness_perturbed , input_complexity_perturbed , intrinsic_complexity_perturbed, import_penalty_perturbed = compute_extended_fitness_with_imports(
            Mcp_export_4_perturbed, Mcp_import_4, C_4.T, alpha = alpha, tensor_window = 3, max_iter = 1000,
            tol = 1e-6, track_convergence = False ,save_delta = False, delta_window = None)

    pearson_coeff_Q, _ = pearsonr(complexity, complexity_perturbed)
    spearman_coeff_Q, _ = spearmanr(complexity, complexity_perturbed)
    pearson_coeff_CQ, _ = pearsonr(input_complexity, input_complexity_perturbed)
    spearman_coeff_CQ, _ = spearmanr(input_complexity, input_complexity_perturbed)
    pearson_coeff_Delta, _ = pearsonr(intrinsic_complexity, intrinsic_complexity_perturbed)
    spearman_coeff_Delta, _ = spearmanr(intrinsic_complexity, intrinsic_complexity_perturbed)
    pearson_coeff_F, _ = pearsonr(fitness, fitness_perturbed)
    spearman_coeff_F, _ = spearmanr(fitness, fitness_perturbed)
    pearson_coeff_S, _ = pearsonr(import_penalty, import_penalty_perturbed)
    spearman_coeff_S, _ = spearmanr(import_penalty, import_penalty_perturbed)

   
    rows.append({
        "eta":              eta,
        "pearson_Q":        pearson_coeff_Q,
        "spearman_Q":       spearman_coeff_Q,
        "pearson_CQ":        pearson_coeff_CQ,
        "spearman_CQ":       spearman_coeff_CQ,
        "pearson_Delta":        pearson_coeff_Delta,
        "spearman_Delta":       spearman_coeff_Delta,
        "pearson_F":        pearson_coeff_F,
        "spearman_F":       spearman_coeff_F,
        "pearson_S":        pearson_coeff_S,
        "spearman_S":       spearman_coeff_S,
    })
```

```python
Mcp_import_stability_table_4_digits = pd.DataFrame(rows)
Mcp_import_stability_table_4_digits.to_parquet(str(ROOT / 'results/stability/Mcp_export/extended_fitness_with_imports/Mcp_export_stability_4_digits.parquet'))
```
