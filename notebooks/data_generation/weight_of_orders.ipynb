# SCAN IN a = ρ·α

Scansione fitta del peso degli ordini al variare di a = ρ α, con ρ autovalore principale di C
(ρ = 1/`compute_alpha(C)`; per C nilpotente `compute_alpha` usa l'in-degree medio).
Stessa griglia di a per 4 e 6 digits, così le curve sono confrontabili: a = 1 è il punto
in cui il termine α C Q ha la stessa scala di Δ.

Output (formato lungo, una riga per a e ordine k):
`results/contributo_ordini/{res}_digits/{C_TYPE}/{ALGO}/scan_a_{res}_digits.parquet`
con colonne `a, alpha, rho, ordine, peso_medio, peso_cumulativo, n_iter`.

```python
from src.algorithms import compute_alpha

A_GRID = np.logspace(-1.5, 1.5, 21)          # 21 punti in [0.03, 31.6], contiene a = 1
K = 11
RESOLUTIONS = [4, 6]
C_FILES = {"C_classica": "C_{res}_digits_HS02.parquet",
           "C_nilpotente": "C_nilpotente_{res}_digits_HS02.parquet"}
C_TYPES = ["C_classica"]
ALGOS = {"extended_fitness_no_imports": compute_extended_fitness_no_imports,
         "extended_eci_no_imports":     compute_extended_eci_no_imports}

DATA = ROOT / "data" / "HS02_all_years"
OUT = ROOT / "results" / "contributo_ordini"
```

```python
for res in RESOLUTIONS:
    base = DATA / f"{res}_digits" / "alligned" / "binarized"
    Mcp_export = pd.read_parquet(base / "Mcp_export" / "Mcp_export_alligned_Y2002.parquet")
    for c_type in C_TYPES:
        C = pd.read_parquet(base / "C" / C_FILES[c_type].format(res=res))
        rho = 1.0 / compute_alpha(C)
        for algo_name, algo in ALGOS.items():
            rows = []
            for a in A_GRID:
                alpha = a / rho
                complexity, fitness, input_complexity, intrinsic_complexity, info, delta_history = algo(
                    Mcp_export, C.T, alpha=alpha, max_iter=5000, tol=1e-6,
                    track_convergence=True, save_delta=True, delta_window=K)
                pesi_df, _ = peso_K_ordini(complexity, delta_history, C, alpha=alpha, K=K)
                pesi_df = pesi_df.assign(a=a, alpha=alpha, rho=rho, n_iter=info["n_iter"])
                rows.append(pesi_df)
                if info["n_iter"] >= 5000 - 1:
                    print(f"[non convergente] {res} digits, {c_type}, {algo_name}, a={a:.3g}")
            scan = pd.concat(rows, ignore_index=True)
            out = OUT / f"{res}_digits" / c_type / algo_name / f"scan_a_{res}_digits.parquet"
            out.parent.mkdir(parents=True, exist_ok=True)
            scan.to_parquet(out)
            print(f"{res} digits, {c_type}, {algo_name}: rho = {rho:.2f}, salvato {out.name}")
```
