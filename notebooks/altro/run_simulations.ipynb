```python
import numpy as np
import pandas as pd
import sys
from pathlib import Path
from scipy.stats import pearsonr, spearmanr

ROOT = next(p for p in [Path().resolve(), *Path().resolve().parents] if (p / "src").is_dir())
sys.path.insert(0, str(ROOT))

from src.algorithms import compute_extended_fitness
from src.mixed_functions import autovettore_principale, make_C_nilpotente

```

```python
C = make_C_nilpotente(C)
```

```python
n_inputs = (C == 1).sum(axis=0) # axis = 0 , l'operazione si muove lungo le righe (corretto)
```

```python
alphas = np.linspace(0,1,101)
```

```python
rows = []

for alpha in alphas:

    complexity, fitness, CQ, Delta = compute_extended_fitness(Mcp,C.T,alpha = alpha, max_iter = 100000, 
                                                              tol = 1e-6, track = False, save_delta = False, delta_window = None)

    autoval_principale, autovett_principale = autovettore_principale(C.T)
    
    coeff_pearson_autovett_Q, _ = pearsonr(autovett_principale, complexity)
    coeff_spearman_autovett_Q, _ = spearmanr(autovett_principale, complexity)
    
    coeff_pearson_autovett_CQ, _ = pearsonr(autovett_principale, CQ)
    coeff_spearman_autovett_CQ, _ = spearmanr(autovett_principale, CQ)

    coeff_pearson_n_inputs_Q, _ = pearsonr(n_inputs, complexity)
    coeff_spearman_n_inputs_Q, _ = spearmanr(n_inputs, complexity)

    coeff_pearson_n_inputs_CQ, _ = pearsonr(n_inputs, CQ)
    coeff_spearman_n_inputs_CQ, _ = spearmanr(n_inputs, CQ)
    
    rows.append({
        "alpha":            alpha,
    
        "pearson_autovett_Q": coeff_pearson_autovett_Q,
        "spearman_autovett_Q": coeff_spearman_autovett_Q,
    
        "pearson_autovett_CQ": coeff_pearson_autovett_CQ,
        "spearman_autovett_CQ": coeff_spearman_autovett_CQ,

        "pearson_n_inputs_Q": coeff_pearson_n_inputs_Q,
        "spearman_n_inputs_Q": coeff_spearman_n_inputs_Q, 

        "pearson_n_inputs_CQ": coeff_pearson_n_inputs_CQ,
        "spearman_n_inputs_CQ": coeff_spearman_n_inputs_CQ,
     })   
```

### FUNZIONI GENERATRICI DI RETI C

```python
# ============================================================
# A) STUDIO IN LUNGHEZZA
# ============================================================

def chain(n):
    """
    Questa funzione genera una catena ORIZZONTALE con n nodi: 
    n-1 -> n-2 -> ... -> 1 -> 0  
    (Il prodotto 0 e' il piu' a valle)

    Parametri:
    - n: numero di nodi della rete

    Output:
    - C : matrice di adiacenza corrispondente a questa rete
    
    """
    
    C = np.zeros((n, n))
    for i in range(n-1):
        C[i, i+1] = 1.0
    return C



def multiple_chain(n, k):
    """
    Questa funzione genera una catena ORIZZONTALE con n nodi, ognuno dei quali da k link a quello successivo
    n-1  k->  n-2  k-> ...  k-> 1   k-> 0  
    (Il prodotto 0 e' il piu' a valle)

    Parametri:
    - n: numero di nodi della rete
    - k: outdegree dei nodi 

    Output:
    - C : matrice di adiacenza corrispondente a questa rete
    
    """
    
    C = np.zeros((n, n))
    for i in range(n-1):
        C[i, i+1] = k
    return C



def chain_with_cycle_last(n, r):
    """
    
    Catena di n nodi + arco 0 -> r 
    (il nodo 0 diventa input del nodo r).
       r=0: nessun ciclo.  r>=1: ciclo di lunghezza r+1 ('ordine r').

    Parametri:
    - n: numero di nodi della rete (collegati a catena orizzontale)
    - r: indice del nodo a cui il primo si collega  (0 --> r)

    Output:
    - C : matrice di adiacenza corrispondente a questa rete
       
       """
    C = chain(n)
    if r >= 1:
        assert r < n
        C[r, 0] = 1.0
    return C



def chain_with_cycle_first(n, r):
    """
    
    Catena di n nodi + arco r -> n 
    (il nodo r diventa input del nodo n).
      

    Parametri:
    - n: numero di nodi della rete (collegati a catena orizzontale)
    - r: indice del nodo a cui il primo si collega  (r --> n)

    Output:
    - C : matrice di adiacenza corrispondente a questa rete
       
       """
    C = chain(n)
    if r < n:
        assert r >= 0
        C[n - 1, r] = 1.0
    return C


# ============================================================
# B) STUDIO IN LARGHEZZA
# ============================================================

def star(m):
    """
    Nodo 0 con m input diretti (ordine 1)

    Parametri:
    - m: numero di nodi in input al nodo 0 (ordine 1)

    Output:
    - C : matrice di adiacenza corrispondente a questa rete
    
    """
    n = 1 + m 
    C = np.zeros((n, n))
    for j in range(1, m+1):
        C[0, j] = 1.0
    return C



def broom(m, depth):
    """
    Serve a confrontare 'largo vicino' vs 'largo lontano'.
    1) Catena 0 <- 1 <- ... <- depth (orizzontale a catena)
    2) Poi m input attaccati al nodo 'depth' (in larghezza a stella)
    
    Parametri:
    - m: numero di nodi in input al nodo depth (ordine depth) , forse potrei sostituire depth ---> n
    - depth: numero di nodi collegati a catena

    Output:
    - C : matrice di adiacenza corrispondente a questa rete
    
    """
    n = depth + 1 + m
    C = np.zeros((n, n))
    for i in range(depth):
        C[i, i+1] = 1.0
    for j in range(depth+1, n):
        C[depth, j] = 1.0
    return C
```

```python
def chain_with_cycle_generalized(n, i, j):
    """
    Catena di n nodi + arco i -> j
    (il nodo i diventa input del nodo j).

    Parametri:
    - n: numero di nodi della rete (collegati a catena orizzontale)
    - i: indice del nodo in input
    - j: indice del nodo in output
    Output:
    - C : matrice di adiacenza corrispondente a questa rete
    """
    if not (0 <= i < n) or not (0 <= j < n):
        raise ValueError(f"Indici fuori dai limiti: i={i}, j={j}, n={n}")
    if i >= j:
        raise ValueError(f"Serve i < j per avere un ciclo (ricevuto i={i}, j={j})")

    C = chain(n)
    C[j, i] = 1.0
    return C
```

## ALGORITMI

```python
# ============================================================
#  MAPPA RIDOTTA  (M = tutti-uno  =>  Delta =  1)
# ============================================================


def compute_second_term(C, max_iter=200000, tol=1e-6):
    """
    Q <- (C Q + 1) / mean(C Q + 1)

    Punto fisso: (mu I - C) Q = 1, mean(Q) = 1.
    mu = fattore di normalizzazione a convergenza; lo sconto per ordine e' 1/mu.
    """
    C = np.asarray(C, float)
    Q = np.ones(C.shape[0])
    for i in range(max_iter):
        Q_old = Q                 # riferimento: sicuro perche' la riga sotto ri-lega Q a un array nuovo
        Q = C @ Q + 1.0
        mu = Q.mean()
        Q /= mu                   # in-place sul nuovo array, Q_old resta intatto
        if np.max(np.abs(Q - Q_old)) < tol:
            return Q, mu, True
    return Q, mu, False
```
