```python
import numpy as np
import pandas as pd
import sys
from pathlib import Path
from scipy.stats import pearsonr, spearmanr

ROOT = next(p for p in [Path().resolve(), *Path().resolve().parents] if (p / "src").is_dir())
sys.path.insert(0, str(ROOT))

from src.algorithms_with_imports import compute_extended_eci_with_imports, compute_extended_fitness_with_imports
from src.C_properties import autovettore_and_autovalore_principale
```

```python
Mcp_export_4 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/4_digits/alligned/binarized/Mcp_export/Mcp_export_alligned_Y2002.parquet'))
Mcp_import_4 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/4_digits/alligned/binarized/Mcp_import/Mcp_import_alligned_Y2002.parquet'))
C_4 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/4_digits/alligned/binarized/C/C_4_digits_HS02.parquet'))
```

```python
Mcp_export_6 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/6_digits/alligned/binarized/Mcp_export/Mcp_export_alligned_Y2002.parquet'))
Mcp_import_6 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/6_digits/alligned/binarized/Mcp_import/Mcp_import_alligned_Y2002.parquet'))
C_6 = pd.read_parquet(str(ROOT / 'data/HS02_all_years/6_digits/alligned/binarized/C/C_6_digits_HS02.parquet'))
```

- RISOLUZIONE: 4 digits

```python
rho, autovettore_principale = autovettore_and_autovalore_principale(C_4.T)
n_inputs = (C_4 == 1).sum(axis=0) # axis = 0 , l'operazione si muove lungo le righe (corretto)
```

```python
lista_a = [0.001, 0.01, 0.05, 0.1,0.2, 0.5, 1, 5, 10, 100]
```

```python
rows = []
```

```python
for a in lista_a:

    alpha = a / rho
    
    complexity, fitness, CQ, Delta, import_penalty = compute_extended_fitness_with_imports(Mcp_export_4, Mcp_import_4, C_4.T, alpha = alpha, tensor_window = 10, max_iter = 1000, 
                                                              tol = 1e-6, track_convergence = False, save_delta = False, delta_window = None)

    coeff_pearson_autovett_Q, _ = pearsonr(autovettore_principale, complexity)
    coeff_spearman_autovett_Q, _ = spearmanr(autovettore_principale, complexity)
    
    coeff_pearson_autovett_CQ, _ = pearsonr(autovettore_principale, CQ)
    coeff_spearman_autovett_CQ, _ = spearmanr(autovettore_principale, CQ)

    coeff_pearson_n_inputs_Q, _ = pearsonr(n_inputs, complexity)
    coeff_spearman_n_inputs_Q, _ = spearmanr(n_inputs, complexity)

    coeff_pearson_n_inputs_CQ, _ = pearsonr(n_inputs, CQ)
    coeff_spearman_n_inputs_CQ, _ = spearmanr(n_inputs, CQ)
    
    rows.append({
        
        "alpha":            alpha,
        "a":            a,
    
        "pearson_autovett_Q": coeff_pearson_autovett_Q,
        "spearman_autovett_Q": coeff_spearman_autovett_Q,
    
        "pearson_autovett_CQ": coeff_pearson_autovett_CQ,
        "spearman_autovett_CQ": coeff_spearman_autovett_CQ,

        "pearson_n_inputs_Q": coeff_pearson_n_inputs_Q,
        "spearman_n_inputs_Q": coeff_spearman_n_inputs_Q, 

        "pearson_n_inputs_CQ": coeff_pearson_n_inputs_CQ,
        "spearman_n_inputs_CQ": coeff_spearman_n_inputs_CQ,
     })   
```

```python
C_correlations_table_4_digits = pd.DataFrame(rows)
C_correlations_table_4_digits.to_parquet(str(ROOT / 'results/correlazioni_autovettore_C/extended_fitness_with_imports/C_correlations_4_digits_controllo.parquet'))
```

# SCAN IN a = ρ·α

Correlazione (Pearson e Spearman, sui prodotti) di Q e CQ con
- l'autovettore principale di C (solo C classica: per C nilpotente tutti gli autovalori sono 0 e l'autovettore non è definito → NaN)
- il numero di input N_1 (somma per colonna di C)

al variare di a = ρ α, con ρ = 1/`compute_alpha(C)` (autovalore principale; in-degree medio per C nilpotente).
Stessa griglia di a per tutte le combinazioni.

Output: `results/correlazioni_autovettore_C/{C_TYPE}/{ALGO}/scan_a_{res}_digits.parquet`

```python
from src.algorithms import compute_alpha
from src.algorithms_with_imports import compute_extended_fitness_with_imports, compute_extended_eci_with_imports

A_GRID = np.logspace(-2, 2, 25)              # 25 punti in [0.01, 100], contiene a = 1
TENSOR_WINDOW = 10
RESOLUTIONS = [4, 6]
C_FILES = {"C_classica": "C_{res}_digits_HS02.parquet",
           "C_nilpotente": "C_nilpotente_{res}_digits_HS02.parquet"}
C_TYPES = ["C_classica", "C_nilpotente"]
ALGOS = {"extended_fitness_with_imports": compute_extended_fitness_with_imports,
         "extended_eci_with_imports":     compute_extended_eci_with_imports}

DATA = ROOT / "data" / "HS02_all_years"
OUT = ROOT / "results" / "correlazioni_autovettore_C"
```

```python
def corr_pair(x, y):
    return pearsonr(x, y)[0], spearmanr(x, y)[0]

def scan_a(res, c_type, algo_name):
    base = DATA / f"{res}_digits" / "alligned" / "binarized"
    Mcp_export = pd.read_parquet(base / "Mcp_export" / "Mcp_export_alligned_Y2002.parquet")
    Mcp_import = pd.read_parquet(base / "Mcp_import" / "Mcp_import_alligned_Y2002.parquet")
    C = pd.read_parquet(base / "C" / C_FILES[c_type].format(res=res))
    rho = 1.0 / compute_alpha(C)
    n_inputs = (C == 1).sum(axis=0)          # numero di input di ogni prodotto (colonna)
    autovett = autovettore_and_autovalore_principale(C.T)[1] if c_type == "C_classica" else None

    rows = []
    for a in A_GRID:
        alpha = a / rho
        complexity, fitness, CQ, Delta, import_penalty, info = ALGOS[algo_name](
            Mcp_export, Mcp_import, C.T, alpha=alpha, tensor_window=TENSOR_WINDOW, max_iter=5000,
            tol=1e-6, track_convergence=True, save_delta=False, delta_window=None)[:6]
        row = {"a": a, "alpha": alpha, "rho": rho, "n_iter": info["n_iter"]}
        for name, X in (("Q", complexity), ("CQ", CQ)):
            row[f"pearson_n_inputs_{name}"], row[f"spearman_n_inputs_{name}"] = corr_pair(n_inputs, X)
            row[f"pearson_autovett_{name}"], row[f"spearman_autovett_{name}"] = (
                corr_pair(autovett, X) if autovett is not None else (np.nan, np.nan))
        rows.append(row)

    out = OUT / c_type / algo_name / f"scan_a_{res}_digits.parquet"
    out.parent.mkdir(parents=True, exist_ok=True)
    pd.DataFrame(rows).to_parquet(out)
    print(f"{res} digits, {c_type}, {algo_name}: rho = {rho:.2f}, salvato {out.relative_to(ROOT)}")
```

```python
for res in RESOLUTIONS:
    for c_type in C_TYPES:
        for algo_name in ALGOS:
            scan_a(res, c_type, algo_name)
```
